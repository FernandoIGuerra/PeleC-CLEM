```python
import numpy as np
import matplotlib.pyplot as plt

def generate_eddy_size(eta, Delta_bar, size=1):
    """
    Generate random eddy sizes l from the PDF:
    f(l) = (5/3 * l**(-8/3)) / (eta**(-5/3) - Delta_bar**(-5/3))
    for eta <= l <= Delta_bar.

    Parameters:
    eta : float - Kolmogorov scale
    Delta_bar : float - LES filter width (integral scale of filter)
    size : int - number of random samples to generate

    Returns:
    numpy.ndarray - array of random eddy sizes
    """
    # Uniform random numbers in [0,1]
    u = np.random.uniform(0, 1, size)

    # Precompute constants
    eta_pow = eta ** (-5.0/3.0)
    Delta_pow = Delta_bar ** (-5.0/3.0)

    # Inverse transform
    X = eta_pow - u * (eta_pow - Delta_pow)
    l = X ** (-3.0/5.0)

    return l

# ============================================
# Example usage
if __name__ == "__main__":
    # Example parameters (adjust as needed)
    eta = 1e-3        # Kolmogorov scale
    Delta_bar = 1.0   # LES filter width (must be > eta)

    # Generate 100,000 samples
    samples = generate_eddy_size(eta, Delta_bar, size=100000)

    # Plot histogram and compare with theoretical PDF
    bins = np.linspace(eta, Delta_bar, 100)
    hist, bin_edges = np.histogram(samples, bins=bins, density=True)

    # Midpoints for plotting
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

    # Theoretical PDF
    denom = eta**(-5/3) - Delta_bar**(-5/3)
    f_theory = (5/3) * bin_centers**(-8/3) / denom

    plt.figure(figsize=(8,5))
    plt.plot(bin_centers, f_theory, 'r-', label='Theoretical PDF', lw=2)
    plt.hist(samples, bins=bins, density=True, alpha=0.6, label='Generated samples')
    plt.xlabel('Eddy size l')
    plt.ylabel('Probability density')
    plt.title('PDF of eddy sizes (Kolmogorov scaling)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    # Print first few samples
    print("First 10 generated eddy sizes:")
    print(samples[:10])
```
