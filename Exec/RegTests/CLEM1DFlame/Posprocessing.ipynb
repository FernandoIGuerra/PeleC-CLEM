```python
def extract_xmom(path):
    """Extract x-momentum along x-axis at domain center"""
    yt.set_log_level(50)
    ds = yt.load(path)
    
    # 1D ray along x at the domain mid-line
    c = ds.domain_center
    ray = ds.ortho_ray(0, (c[1], c[2]))
    order = np.argsort(ray["boxlib", "x"])
    
    x = np.array(ray["boxlib", "x"][order])
    xmom = np.array(ray["boxlib", "xmom"][order])  # x-momentum
    pressure = np.array(ray["boxlib", "press"][order])  # pressure
    velocity = np.array(ray["boxlib", "velocity_x"][order])  # x-velocity
    
    return x, xmom, pressure, velocity


x_n18, xmom_n18 = extract_xmom("Iso_isentropic/plt_01000")
x, xmom = extract_xmom("/home/fernando/FPCE/PeleC/Exec/RegTests/1DFlame/CurrentResultsNR/pltFlameNR_01000")
# Compare

# Plot comparison
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
#plt.plot(x_n9, xmom_n9, label='N9', linewidth=1)
plt.plot(x_n18, xmom_n18, label='N18', linewidth=1)
#plt.plot(x_n27, xmom_n27, label='N27', linewidth=2)
#plt.plot(x_n36, xmom_n36, label='N36', linewidth=2,)
plt.plot(x, xmom, label='Ref', linewidth=2, marker = "s", markevery = 5, linestyle = ":")
plt.xlabel('x position')
plt.ylabel('x-momentum')
plt.legend()
plt.grid(True)
plt.title('Comparison of x-momentum: N18 vs N27')
plt.show()
```
