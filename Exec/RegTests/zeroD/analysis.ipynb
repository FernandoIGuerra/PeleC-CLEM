```python
import matplotlib.pyplot as plt
import yt
import numpy as np
import os

root_PeleC = "/home/fernando/FPCE/PeleC/Exec/RegTests/zeroD"
species = ["H2","O2","H2O","OH","H","O","HO2","H2O2","N2"]
```
